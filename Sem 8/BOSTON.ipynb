```python
# Boston House Prediction - Linear Regression
import pandas as pd
from sklearn.linear_model import LinearRegression

d=pd.read_csv("BostonHousing.csv")

x=d.drop("medv",axis=1)
y=d["medv"]

m=LinearRegression().fit(x,y)

print(m.predict([x.iloc[0]]))
```

```python
# Boston House Prediction - DNN
import pandas as pd
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

d=pd.read_csv("BostonHousing.csv")

x=d.drop("medv",axis=1)
y=d["medv"]

m=Sequential([Dense(64,activation='relu'),Dense(1)])

m.compile("adam","mse")

m.fit(x,y,epochs=5,verbose=0)

print(m.predict(x[:1]))
```
